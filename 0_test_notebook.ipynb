```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict 
```

```python
class BMIState(TypedDict):
    weight: float
    height: float
    bmi: float
    verdict: str
```

```python
#define a simple workflow 
graph = StateGraph(BMIState) 
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    weight = state.get("weight")
    height = state.get("height")
    if weight is None or height is None:
        raise ValueError("Weight and height must be provided.")
    bmi = weight / (height ** 2)
    state["bmi"] = bmi
    return state
```

```python
def label_bmi(state: BMIState) -> BMIState:
    bmi = state.get("bmi")
    if bmi is None:
        raise ValueError("BMI must be calculated before labeling.")
    if bmi < 18.5:
        state["label"] = "Underweight"
    elif 18.5 <= bmi < 24.9:
        state["label"] = "Normal weight"
    elif 25 <= bmi < 29.9:
        state["label"] = "Overweight"
    else:
        state["label"] = "Obesity"
    return state
```

```python
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('label_bmi', label_bmi)
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'label_bmi')
graph.add_edge('label_bmi', END)
workflow = graph.compile()
```

```python
intial_state = BMIState(weight=75, height=1.75)
final_state = workflow.invoke(intial_state) 

print(final_state)
```
